```python

import os
import urllib.request

# Create weights directory
weights_dir = os.path.join(os.getcwd(), "weights")
os.makedirs(weights_dir, exist_ok=True)

# Official YOLO26 weight download URLs
base_url = "https://github.com/ultralytics/assets/releases/download/v8.3.0/"

models = [
    "yolo26n.pt",
    "yolo26s.pt",
    "yolo26m.pt",
    "yolo26l.pt",
    "yolo26x.pt"
]

# Download each model
for model in models:

    url = base_url + model
    file_path = os.path.join(weights_dir, model)

    if os.path.exists(file_path):
        print(f"Already exists: {model}")
        continue

    print(f"Downloading {model}...")

    try:
        urllib.request.urlretrieve(url, file_path)
        print(f"Downloaded: {file_path}")

    except Exception as e:
        print(f"Failed to download {model}: {e}")

print("YOLO26 weight download process completed!")
```
